<font size="4" color='#CC403E'><strong>✔️ 2. O que é um Data Lake?</strong></font>

<font size="2" color='#66666'>

>- Um **Data Lake é uma arquitetura de armazenamento voltada para guardar grandes volumes de dados em formatos variados**.

>- Ele pode armazenar:
    - *arquivos CSV;*
    - *arquivos JSON;*
    - *arquivos Parquet;*
    - *logs;*
    - *imagens;*
    - *dados semiestruturados;*
    - *dados ainda não tratados.*

>- **Visão prática**

>- No contexto deste projeto, o Data Lake representa a área onde os dados brutos de crédito chegam antes de qualquer tratamento.

>- Exemplo:
    - cadastro de clientes;
    - contratos;
    - pagamentos;
    - atrasos;
    - transações;
    - histórico de crédito;
    - eventos de cobrança.

---

</font></div>

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 3. O que é um Data Warehouse?</strong></font>

<font size="2" color='#66666'>

>- Um **Data Warehouse é uma arquitetura orientada a dados estruturados, modelados e preparados para análise**.

>- Ele normalmente é usado para:
    - *relatórios executivos;*
    - *dashboards;*
    - *indicadores de negócio;*
    - *consultas SQL;*
    - *análises históricas;*
    - *dados já tratados e confiáveis.*

>- **Visão prática**

>- No contexto de risco de crédito, o Data Warehouse seria usado para responder perguntas como:
    - Qual é a taxa de inadimplência por segmento?
    - Qual é o volume de crédito por faixa de renda?
    - Qual é o percentual de atraso por safra?
    - Quais clientes possuem maior exposição ao risco?

---

</font></div>

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 4. O que é um Lakehouse?</strong></font>

<font size="2" color='#66666'>

>- **Lakehouse é uma arquitetura que combina características de Data Lake e Data Warehouse**.

>- A proposta é permitir que dados brutos, tratados e analíticos convivam em uma arquitetura única, com suporte a:
    - *armazenamento escalável;*
    - *transações;*
    - *tabelas confiáveis;*
    - *processamento distribuído;*
    - *SQL;*
    - *machine learning;*
    - *governança;*
    - *versionamento;*
    - *pipelines analíticos.*

>- **Interpretação simples**

>- O Lakehouse tenta resolver uma dor comum:
    - O Data Lake é flexível, mas pode virar bagunça.  
    - O Data Warehouse é organizado, mas pode ser caro e menos flexível.  
    - O Lakehouse busca unir flexibilidade com governança e confiabilidade.

---

</font></div>

<font size="4" color='#CC403E'><strong>✔️ 6. O que é Delta Lake?</strong></font>

<font size="2" color='#66666'>

>- **Delta Lake é uma camada de armazenamento transacional usada em arquiteturas Lakehouse**.

>- Ele adiciona recursos importantes sobre arquivos de dados, como:
    - *transações ACID;*
    - *versionamento;*
    - *histórico de alterações;*
    - *atualização e deleção de registros;*
    - *suporte a operações incrementais;*
    - *controle de schema;*
    - *maior confiabilidade para pipelines.*

>- **Por que isso importa?**

>- Em projetos de risco de crédito, confiabilidade dos dados é crítica.

>- Não basta apenas armazenar arquivos. É necessário controlar:
    - quais dados entraram;
    - quando entraram;
    - se foram corrigidos;
    - qual versão está sendo analisada;
    - se a tabela pode ser reproduzida;
    - se o pipeline é auditável.

</font></div>

<font size="4" color='#CC403E'><strong>✔️ 10. Diferença entre Managed Table e External Table</strong></font>

<font size="2" color='#66666'>

>- Em Databricks, é importante entender a diferença conceitual entre tabelas gerenciadas e externas.

>- **Managed Table**
    - A plataforma gerencia os dados e os metadados da tabela.

>- **External Table**
    - A plataforma gerencia os metadados, mas os dados ficam em uma localização externa controlada pelo usuário ou pela organização.

>- **Interpretação prática**
    - Para fins de estudo, podemos começar com tabelas gerenciadas.  
    - Para cenários corporativos, tabelas externas podem ser importantes quando há necessidade de controle explícito sobre armazenamento, paths e integração com Data Lake corporativo.

---

</font></div>
