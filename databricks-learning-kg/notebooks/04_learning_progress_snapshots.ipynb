</font></div>

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 2. Atualização dos scores</strong></font>

<font size="2" color='#66666'>
    
>- Agora aplicamos os eventos de aprendizagem aos tópicos existentes.

>- Os scores são limitados a 1.0 para manter uma escala interpretável:
    - `confidence_score`: percepção de domínio conceitual;
    - `mastery_score`: evidência prática acumulada.


</font></div>

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 2.1. Atualização tópicos</strong></font>

<font size="2" color='#66666'></font></div>

```python
progress_deltas = (
    learning_events
    .groupby("topic_id", as_index=False)
    .agg(
        confidence_delta=("confidence_delta", "sum"),
        mastery_delta=("mastery_delta", "sum"),
        evidence_count=("event_id", "count"),
    )
)

topics_progress = topics.merge(
    progress_deltas,
    on="topic_id",
    how="left",
)

topics_progress[["confidence_delta", "mastery_delta", "evidence_count"]] = (
    topics_progress[["confidence_delta", "mastery_delta", "evidence_count"]]
    .fillna(0)
)

topics_progress["previous_confidence_score"] = topics_progress["confidence_score"]
topics_progress["previous_mastery_score"] = topics_progress["mastery_score"]

topics_progress["confidence_score"] = (
    topics_progress["confidence_score"] + topics_progress["confidence_delta"]
).clip(upper=1.0)

topics_progress["mastery_score"] = (
    topics_progress["mastery_score"] + topics_progress["mastery_delta"]
).clip(upper=1.0)

topics_progress["last_review_date"] = date.today().isoformat()

topics_progress["status"] = topics_progress["mastery_score"].apply(
    lambda score: (
        "not_started" if score == 0
        else "mapped" if score < 0.25
        else "practicing" if score < 0.50
        else "consolidating" if score < 0.75
        else "strong"
    )
)

topics_progress.head()
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 2.2. Ver progresso aplicado</strong></font>

<font size="2" color='#66666'></font></div>

```python
progress_view = topics_progress[
    [
        "topic_id",
        "name",
        "category",
        "priority",
        "previous_confidence_score",
        "confidence_score",
        "previous_mastery_score",
        "mastery_score",
        "evidence_count",
        "status",
    ]
].sort_values(
    by=["evidence_count", "mastery_score"],
    ascending=False,
)

progress_view.head(20)
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 3. Snapshot da Semana 01</strong></font>

<font size="2" color='#66666'>
    
>- Agora criamos o primeiro snapshot evolutivo após os primeiros notebooks.

>- Esse snapshot representa o estado da jornada depois de:
    - Notebook 00;
    - Notebook 01;
    - Notebook 02;
    - Notebook 03.

</font></div>

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 4. Criar snapshot semanal</strong></font>

<font size="2" color='#66666'></font></div>

```python
week_number = 1
snapshot_id = "SNAP_W01"

snapshot_w01 = pd.DataFrame(
    [
        {
            "snapshot_id": snapshot_id,
            "snapshot_date": date.today().isoformat(),
            "week_number": week_number,
            "overall_progress": round(topics_progress["mastery_score"].mean(), 4),
            "coverage_pct": round((topics_progress["mastery_score"] > 0).mean(), 4),
            "avg_confidence": round(topics_progress["confidence_score"].mean(), 4),
            "avg_mastery": round(topics_progress["mastery_score"].mean(), 4),
            "main_gap": "Prática real em Databricks Workflows, Delta Lake, Auto Loader e Unity Catalog ainda precisa ser aprofundada.",
            "main_achievement": "Primeira versão operacional do Knowledge Graph criada, visualizada e carregada no Neo4j.",
        }
    ]
)

snapshot_w01
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 5. Snapshot por tópico</strong></font>

<font size="2" color='#66666'></font></div>

```python
topic_progress_snapshot_w01 = topics_progress[
    [
        "topic_id",
        "name",
        "domain_id",
        "category",
        "priority",
        "confidence_score",
        "mastery_score",
        "status",
        "last_review_date",
    ]
].copy()

topic_progress_snapshot_w01.insert(0, "snapshot_id", snapshot_id)
topic_progress_snapshot_w01.insert(1, "snapshot_date", date.today().isoformat())
topic_progress_snapshot_w01.insert(2, "week_number", week_number)

topic_progress_snapshot_w01.head()
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 6. Comparativo inicial</strong></font>

<font size="2" color='#66666'></font></div>

```python
comparison = topics.merge(
    topics_progress[
        [
            "topic_id",
            "confidence_score",
            "mastery_score",
            "status",
        ]
    ],
    on="topic_id",
    suffixes=("_w00", "_w01"),
)

comparison["confidence_gain"] = (
    comparison["confidence_score_w01"] - comparison["confidence_score_w00"]
).round(4)

comparison["mastery_gain"] = (
    comparison["mastery_score_w01"] - comparison["mastery_score_w00"]
).round(4)

comparison_view = comparison[
    [
        "topic_id",
        "name",
        "category",
        "priority",
        "confidence_score_w00",
        "confidence_score_w01",
        "confidence_gain",
        "mastery_score_w00",
        "mastery_score_w01",
        "mastery_gain",
        "status_w01",
    ]
].sort_values(
    by=["mastery_gain", "confidence_gain"],
    ascending=False,
)

comparison_view.head(20)
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 7. Leitura analítica do progresso</strong></font>

<font size="2" color='#66666'>
    
>- Agora vamos identificar:
    - tópicos com maior evolução;
    - tópicos ainda sem evidência prática;
    - tópicos prioritários ainda frágeis;
    - média de progresso por domínio.

</font></div>

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 7.1. Tópico com maior ganho</strong></font>

<font size="2" color='#66666'></font></div>

```python
top_progress_topics = comparison_view[
    (comparison_view["confidence_gain"] > 0) | (comparison_view["mastery_gain"] > 0)
].head(15)

top_progress_topics
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 7.2. Tópico prioritários ainda frágeis</strong></font>

<font size="2" color='#66666'></font></div>

```python
weak_high_priority_topics = topics_progress[
    (topics_progress["priority"] == "high") &
    (topics_progress["mastery_score"] < 0.30)
][
    [
        "topic_id",
        "name",
        "category",
        "priority",
        "confidence_score",
        "mastery_score",
        "status",
    ]
].sort_values(
    by=["mastery_score", "confidence_score"],
    ascending=True,
)

weak_high_priority_topics.head(20)
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 7.3. Progresso por domínio</strong></font>

<font size="2" color='#66666'></font></div>

```python
domain_progress = (
    topics_progress
    .merge(exam_domains[["domain_id", "name", "weight_pct"]], on="domain_id", how="left", suffixes=("_topic", "_domain"))
    .groupby(["domain_id", "name_domain", "weight_pct"], as_index=False)
    .agg(
        total_topics=("topic_id", "count"),
        avg_confidence=("confidence_score", "mean"),
        avg_mastery=("mastery_score", "mean"),
        topics_with_evidence=("mastery_score", lambda s: (s > 0).sum()),
    )
)

domain_progress["coverage_pct"] = (
    domain_progress["topics_with_evidence"] / domain_progress["total_topics"]
).round(4)

domain_progress["avg_confidence"] = domain_progress["avg_confidence"].round(4)
domain_progress["avg_mastery"] = domain_progress["avg_mastery"].round(4)

domain_progress.sort_values("weight_pct", ascending=False)
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 8. Exportação dos artefatos de progresso</strong></font>

<font size="2" color='#66666'>
    
>- Vamos salvar os arquivos gerados para uso posterior no Neo4j, no dashboard e no README do projeto.

</font></div>

```python
learning_events.to_csv(PROCESSED_DIR / "learning_events.csv", index=False, encoding="utf-8")
topics_progress.to_csv(PROCESSED_DIR / "topics_progress_w01.csv", index=False, encoding="utf-8")
snapshot_w01.to_csv(PROCESSED_DIR / "snapshot_w01.csv", index=False, encoding="utf-8")
topic_progress_snapshot_w01.to_csv(PROCESSED_DIR / "topic_progress_snapshot_w01.csv", index=False, encoding="utf-8")

comparison_view.to_csv(EXPORTS_DIR / "learning_progress_comparison_w00_w01.csv", index=False, encoding="utf-8")
top_progress_topics.to_csv(EXPORTS_DIR / "top_progress_topics_w01.csv", index=False, encoding="utf-8")
weak_high_priority_topics.to_csv(EXPORTS_DIR / "weak_high_priority_topics_w01.csv", index=False, encoding="utf-8")
domain_progress.to_csv(EXPORTS_DIR / "domain_progress_w01.csv", index=False, encoding="utf-8")

print("Arquivos de progresso exportados com sucesso.")
```
