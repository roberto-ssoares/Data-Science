```python
PROJECT_NAME = "Databricks Certification Learning Knowledge Graph"

BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

DATA_DIR = BASE_DIR / "data"
PROCESSED_DIR = DATA_DIR / "processed"
EXPORTS_DIR = DATA_DIR / "exports"
ASSETS_DIR = BASE_DIR / "assets"
DOCS_DIR = BASE_DIR / "docs"

for directory in [PROCESSED_DIR, EXPORTS_DIR, ASSETS_DIR, DOCS_DIR]:
    directory.mkdir(parents=True, exist_ok=True)

BASE_DIR
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 1. Carregamento dos dados do grafo</strong></font>

<font size="2" color='#66666'>

>- Nesta etapa, vamos carregar os arquivos consolidados criados no Notebook 01:
    - `nodes_all.csv`
    - `edges_all.csv`

>- Esses arquivos representam, respectivamente, os nós e os relacionamentos do Knowledge Graph.

---

</font></div>

```python
nodes_path = PROCESSED_DIR / "nodes_all.csv"
edges_path = PROCESSED_DIR / "edges_all.csv"

nodes_all = pd.read_csv(nodes_path)
edges_all = pd.read_csv(edges_path)

display(nodes_all.head())
display(edges_all.head())
```

```python
print(f"Total de nós: {len(nodes_all)}")
print(f"Total de relacionamentos: {len(edges_all)}")

display(nodes_all["node_label"].value_counts().rename_axis("node_label").reset_index(name="total_nodes"))
display(edges_all["relationship"].value_counts().rename_axis("relationship").reset_index(name="total_edges"))
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 2. Validação antes da construção do grafo</strong></font>

<font size="2" color='#66666'>

>- Antes de criar o grafo, vamos validar:
    - se todos os nós possuem identificador;
    - se todos os relacionamentos possuem origem e destino;
    - se não existem relacionamentos apontando para nós inexistentes.

---

</font></div>

```python
required_node_columns = ["node_id", "node_label", "name"]
required_edge_columns = ["source_id", "relationship", "target_id"]

missing_node_columns = [col for col in required_node_columns if col not in nodes_all.columns]
missing_edge_columns = [col for col in required_edge_columns if col not in edges_all.columns]

assert not missing_node_columns, f"Colunas ausentes em nodes_all: {missing_node_columns}"
assert not missing_edge_columns, f"Colunas ausentes em edges_all: {missing_edge_columns}"

assert nodes_all["node_id"].notna().all(), "Existem node_id nulos."
assert edges_all["source_id"].notna().all(), "Existem source_id nulos."
assert edges_all["target_id"].notna().all(), "Existem target_id nulos."

node_ids = set(nodes_all["node_id"].astype(str))

orphan_sources = sorted(set(edges_all["source_id"].astype(str)) - node_ids)
orphan_targets = sorted(set(edges_all["target_id"].astype(str)) - node_ids)

assert not orphan_sources, f"Existem source_ids órfãos: {orphan_sources}"
assert not orphan_targets, f"Existem target_ids órfãos: {orphan_targets}"

print("Validação concluída com sucesso.")
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 3. Construção do Knowledge Graph com NetworkX</strong></font>

<font size="2" color='#66666'>

>- Vamos construir duas representações:
    1. `kg_multi`: grafo completo do tipo `MultiDiGraph`, preservando múltiplos relacionamentos entre os mesmos nós.
    2. `kg_simple`: projeção simplificada do grafo, usada para métricas de centralidade e conectividade.

---

</font></div>

```python
def clean_attr(value):
    """
    Converte valores nulos em string vazia para evitar problemas na visualização.
    """
    if pd.isna(value):
        return ""
    return value


kg_multi = nx.MultiDiGraph(name=PROJECT_NAME)

for _, row in nodes_all.iterrows():
    node_id = str(row["node_id"])
    
    attrs = {
        col: clean_attr(row[col])
        for col in nodes_all.columns
        if col != "node_id"
    }
    
    kg_multi.add_node(node_id, **attrs)

for _, row in edges_all.iterrows():
    source_id = str(row["source_id"])
    target_id = str(row["target_id"])
    
    attrs = {
        col: clean_attr(row[col])
        for col in edges_all.columns
        if col not in ["source_id", "target_id"]
    }
    
    kg_multi.add_edge(
        source_id,
        target_id,
        key=str(row.get("edge_id", "")),
        **attrs,
    )

print(kg_multi)
```

```python
kg_simple = nx.DiGraph(name=f"{PROJECT_NAME} - Simple Projection")

for node_id, attrs in kg_multi.nodes(data=True):
    kg_simple.add_node(node_id, **attrs)

for source_id, target_id, attrs in kg_multi.edges(data=True):
    relationship = attrs.get("relationship", "")
    
    if kg_simple.has_edge(source_id, target_id):
        existing_relationships = kg_simple[source_id][target_id].get("relationships", [])
        existing_relationships.append(relationship)
        kg_simple[source_id][target_id]["relationships"] = sorted(set(existing_relationships))
        kg_simple[source_id][target_id]["weight"] += 1
    else:
        kg_simple.add_edge(
            source_id,
            target_id,
            relationships=[relationship],
            weight=1,
        )

print(kg_simple)
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 4. Métricas estruturais do grafo</strong></font>

<font size="2" color='#66666'>

>- Agora vamos calcular métricas iniciais:
    - *quantidade de nós;*
    - *quantidade de relacionamentos;*
    - *densidade;*
    - *componentes fracamente conectados;*
    - *grau de entrada;*
    - *grau de saída;*
    - *grau total;*
    - *centralidade de grau.*

>- Essas métricas ajudam a entender quais entidades estão mais conectadas dentro do Knowledge Graph.

---

</font></div>

```python
graph_summary = {
    "project": PROJECT_NAME,
    "nodes": kg_multi.number_of_nodes(),
    "edges_multi": kg_multi.number_of_edges(),
    "edges_simple": kg_simple.number_of_edges(),
    "density_simple": round(nx.density(kg_simple), 6),
    "weakly_connected_components": nx.number_weakly_connected_components(kg_simple),
    "is_weakly_connected": nx.is_weakly_connected(kg_simple),
}

pd.DataFrame([graph_summary])
```

```python
degree_centrality = nx.degree_centrality(kg_simple)
in_degree = dict(kg_simple.in_degree())
out_degree = dict(kg_simple.out_degree())
total_degree = dict(kg_simple.degree())

graph_metrics = pd.DataFrame(
    [
        {
            "node_id": node_id,
            "name": kg_simple.nodes[node_id].get("name", ""),
            "node_label": kg_simple.nodes[node_id].get("node_label", ""),
            "in_degree": in_degree.get(node_id, 0),
            "out_degree": out_degree.get(node_id, 0),
            "total_degree": total_degree.get(node_id, 0),
            "degree_centrality": round(degree_centrality.get(node_id, 0), 6),
        }
        for node_id in kg_simple.nodes()
    ]
).sort_values(
    by=["total_degree", "degree_centrality"],
    ascending=False,
)

graph_metrics.head(15)
```

```python
top_connected_nodes = graph_metrics.head(20)

display(top_connected_nodes)
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 5. Leitura analítica inicial</strong></font>

<font size="2" color='#66666'>

>- Nesta etapa, analisamos os nós mais conectados do KG.

>- Em um Knowledge Graph de aprendizagem, nós com maior conectividade tendem a representar:
    - tópicos centrais;
    - domínios relevantes;
    - artefatos que conectam várias áreas;
    - conceitos estruturantes da jornada.

---

</font></div>

```python
label_summary = (
    graph_metrics
    .groupby("node_label", as_index=False)
    .agg(
        total_nodes=("node_id", "count"),
        avg_total_degree=("total_degree", "mean"),
        max_total_degree=("total_degree", "max"),
        avg_centrality=("degree_centrality", "mean"),
    )
    .sort_values("avg_total_degree", ascending=False)
)

label_summary
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 6. Análise dos tópicos mais conectados</strong></font>

<font size="2" color='#66666'>

>- Agora vamos filtrar apenas os nós do tipo `Topic`.

>- Isso nos ajuda a enxergar quais temas aparecem como mais centrais na ontologia inicial.

---

</font></div>

```python
topic_metrics = (
    graph_metrics
    .query("node_label == 'Topic'")
    .sort_values(["total_degree", "degree_centrality"], ascending=False)
)

topic_metrics.head(15)
```

```python
topic_metrics_path = EXPORTS_DIR / "topic_graph_metrics.csv"
graph_metrics_path = EXPORTS_DIR / "graph_metrics.csv"
label_summary_path = EXPORTS_DIR / "label_summary.csv"

topic_metrics.to_csv(topic_metrics_path, index=False, encoding="utf-8")
graph_metrics.to_csv(graph_metrics_path, index=False, encoding="utf-8")
label_summary.to_csv(label_summary_path, index=False, encoding="utf-8")

topic_metrics_path, graph_metrics_path, label_summary_path
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>✔️ 7. Visualização interativa com PyVis</strong></font>

<font size="2" color='#66666'>

>- Nesta etapa, vamos gerar uma visualização HTML navegável do Knowledge Graph.

>- A visualização será salva em:
    - `assets/databricks_learning_kg_initial_graph.html`

>- Essa versão servirá como primeira inspeção visual do grafo.

---

</font></div>

```python
LABEL_COLORS = {
    "Certification": "#1f77b4",
    "ExamDomain": "#ff7f0e",
    "Topic": "#2ca02c",
    "Subtopic": "#98df8a",
    "Resource": "#9467bd",
    "Notebook": "#8c564b",
    "Lab": "#e377c2",
    "Skill": "#7f7f7f",
    "StudySession": "#bcbd22",
    "Snapshot": "#17becf",
}


def node_size(node_id: str) -> int:
    """
    Define o tamanho visual do nó com base no grau total.
    """
    degree = total_degree.get(node_id, 0)
    return 12 + min(degree * 3, 35)


net = Network(
    height="750px",
    width="100%",
    directed=True,
    notebook=False,
    cdn_resources="in_line",
)

net.barnes_hut(
    gravity=-4500,
    central_gravity=0.25,
    spring_length=160,
    spring_strength=0.03,
    damping=0.85,
)

for node_id, attrs in kg_simple.nodes(data=True):
    label = attrs.get("node_label", "")
    name = attrs.get("name", node_id)
    category = attrs.get("category", "")
    status = attrs.get("status", "")
    priority = attrs.get("priority", "")
    
    title = f"""
    <b>{name}</b><br>
    ID: {node_id}<br>
    Label: {label}<br>
    Category: {category}<br>
    Status: {status}<br>
    Priority: {priority}<br>
    Degree: {total_degree.get(node_id, 0)}
    """
    
    net.add_node(
        node_id,
        label=name,
        title=title,
        color=LABEL_COLORS.get(label, "#cccccc"),
        size=node_size(node_id),
        group=label,
    )

for source_id, target_id, attrs in kg_simple.edges(data=True):
    relationships = attrs.get("relationships", [])
    relationship_label = ", ".join(relationships)
    
    net.add_edge(
        source_id,
        target_id,
        title=relationship_label,
        label=relationship_label if len(relationship_label) <= 22 else "",
        arrows="to",
    )

net.show_buttons(filter_=["physics"])

graph_html_path = ASSETS_DIR / "databricks_learning_kg_initial_graph.html"

# Correção para Windows:
# Em vez de usar net.write_html(), geramos o HTML e salvamos explicitamente em UTF-8.
html_content = net.generate_html(notebook=False)

graph_html_path.write_text(html_content, encoding="utf-8")

graph_html_path
```
