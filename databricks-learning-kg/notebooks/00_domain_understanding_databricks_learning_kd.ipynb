<font size="4" color='#CC403E'><strong>✔️ 7. Modelo inicial de progresso</strong></font>

<font size="2" color='#66666'>

>- O KG acompanhará dois tipos de evolução:

>- **Confidence Score**
    - Mede a percepção subjetiva de segurança sobre o tópico.

>- **Mastery Score**
    - Mede evidência prática associada ao tópico.

>- Essa separação é importante porque um tópico pode parecer conhecido, mas ainda não ter evidência prática suficiente.

---

</font></div>

```python
progress_scale = pd.DataFrame(
    [
        {"score": 0.00, "confidence_meaning": "Não conheço", "mastery_meaning": "Sem evidência prática"},
        {"score": 0.25, "confidence_meaning": "Já li sobre", "mastery_meaning": "Resumo criado"},
        {"score": 0.50, "confidence_meaning": "Entendo conceitualmente", "mastery_meaning": "Notebook executado"},
        {"score": 0.75, "confidence_meaning": "Já pratiquei", "mastery_meaning": "Lab completo"},
        {"score": 1.00, "confidence_meaning": "Consigo explicar e aplicar", "mastery_meaning": "Projeto publicado ou explicado"},
    ]
)

progress_scale
```

<font size="4" color='#CC403E'><strong>✔️ 8. Primeira leitura estratégica</strong></font>

<font size="2" color='#66666'>

>- Neste momento, os tópicos com maior peso e maior impacto prático são:
    - *Development and Ingestion*
    - *Data Processing & Transformations*
    - *Productionizing Data Pipelines*

>- Esses domínios serão priorizados porque concentram a maior parte do exame e também geram evidências fortes para portfólio.

</font></div>

```python
priority_view = (
    exam_domains
    .sort_values(by="weight_pct", ascending=False)
    .assign(
        strategic_focus=lambda df: df["weight_pct"].apply(
            lambda x: "primary" if x >= 18 else "support"
        )
    )
)

priority_view
```
