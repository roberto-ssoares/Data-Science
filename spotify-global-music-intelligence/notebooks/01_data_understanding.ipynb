<font size="3" color='#CC403E'><strong>✔️ Carregamento inicial dos arquivos CSV</strong></font>

<font size="2" color='#66666'></font></div>

```python
def read_csv_with_fallback(file_path: Path) -> pd.DataFrame:
    """
    Lê um arquivo CSV usando tentativas de encoding e separador.

    A função tenta combinações comuns para reduzir erros de leitura,
    especialmente em datasets públicos baixados de plataformas externas.
    """
    attempts = [
        {"encoding": "utf-8", "sep": ","},
        {"encoding": "utf-8-sig", "sep": ","},
        {"encoding": "latin1", "sep": ","},
        {"encoding": "utf-8", "sep": ";"},
        {"encoding": "utf-8-sig", "sep": ";"},
        {"encoding": "latin1", "sep": ";"},
    ]

    last_error = None

    for params in attempts:
        try:
            return pd.read_csv(file_path, low_memory=False, **params)
        except Exception as error:
            last_error = error

    raise RuntimeError(
        f"Não foi possível ler o arquivo {file_path}. "
        f"Último erro encontrado: {last_error}"
    )
    
```

```python
df_spotify_modern = read_csv_with_fallback(SPOTIFY_MODERN_FILE)
df_spotify_tracks = read_csv_with_fallback(SPOTIFY_TRACKS_FILE)

print("Datasets carregados com sucesso.")
```

<div style="background-color:#f3f2ee">
    
<font size="3" color='#CC403E'><strong>✔️ Criação de período analítico para análise temporal</strong></font>

<font size="2" color='#66666'></font></div>

```python
def classify_release_period(year: float) -> str:
    """
    Classifica o ano de lançamento em períodos analíticos.
    """
    if pd.isna(year):
        return "unknown"

    year = int(year)

    if year < 2009:
        return "catalogo_historico_pre_2009"
    if 2009 <= year <= 2015:
        return "streaming_early_2009_2015"
    if 2016 <= year <= 2020:
        return "streaming_growth_2016_2020"
    if 2021 <= year <= 2023:
        return "streaming_mature_2021_2023"
    if 2024 <= year <= 2025:
        return "recent_releases_2024_2025"

    return "out_of_scope"


df_main["periodo_analitico"] = df_main["release_year"].apply(classify_release_period)

period_summary = (
    df_main
    .groupby("periodo_analitico", dropna=False)
    .agg(
        track_count=("track_id", "count"),
        min_year=("release_year", "min"),
        max_year=("release_year", "max"),
        avg_track_popularity=("track_popularity", "mean"),
        median_track_popularity=("track_popularity", "median"),
    )
    .reset_index()
    .sort_values("min_year")
)

period_summary
```

```python
duplicate_key = ["track_name", "artist_name", "album_name"]

text_duplicate_groups = (
    df_main
    .groupby(duplicate_key, dropna=False)
    .agg(
        group_size=("track_id", "count"),
        unique_track_ids=("track_id", "nunique"),
        min_release_date=("album_release_date_parsed", "min"),
        max_release_date=("album_release_date_parsed", "max"),
        min_popularity=("track_popularity", "min"),
        max_popularity=("track_popularity", "max"),
        avg_popularity=("track_popularity", "mean"),
        min_duration=("track_duration_min", "min"),
        max_duration=("track_duration_min", "max"),
    )
    .reset_index()
)

text_duplicate_groups = text_duplicate_groups[text_duplicate_groups["group_size"] > 1].copy()

text_duplicate_groups["popularity_range"] = (
    text_duplicate_groups["max_popularity"] - text_duplicate_groups["min_popularity"]
)

text_duplicate_groups["duration_range"] = (
    text_duplicate_groups["max_duration"] - text_duplicate_groups["min_duration"]
)

text_duplicate_groups.sort_values(
    ["group_size", "popularity_range"],
    ascending=[False, False]
).head(30)
```
