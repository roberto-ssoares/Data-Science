```python
def read_csv_with_fallback(file_path: Path) -> pd.DataFrame:
    """
    Lê um arquivo CSV usando tentativas de encoding e separador.
    """
    attempts = [
        {"encoding": "utf-8", "sep": ","},
        {"encoding": "utf-8-sig", "sep": ","},
        {"encoding": "latin1", "sep": ","},
        {"encoding": "utf-8", "sep": ";"},
        {"encoding": "utf-8-sig", "sep": ";"},
        {"encoding": "latin1", "sep": ";"},
    ]

    last_error = None

    for params in attempts:
        try:
            return pd.read_csv(file_path, low_memory=False, **params)
        except Exception as error:
            last_error = error

    raise RuntimeError(
        f"Não foi possível ler o arquivo {file_path}. "
        f"Último erro encontrado: {last_error}"
    )


df_raw = read_csv_with_fallback(SPOTIFY_MAIN_FILE)

print("Base principal carregada com sucesso.")
print(f"Linhas: {df_raw.shape[0]:,}")
print(f"Colunas: {df_raw.shape[1]:,}")
```

<font size="4" color='#CC403E'><strong>📌 10. Criação do período analítico</strong></font>

<font size="2" color='#66666'>

>- Classificar as faixas em períodos analíticos para apoiar comparações temporais na EDA.

>- **Ações realizadas**

>- **Justificativa técnica**
    - A base possui registros entre 1952 e 2025.
    - Como o projeto tem foco em tendências musicais e popularidade, a criação de períodos ajuda a comparar:

<div style="background-color:#f3f2ee">
    
<font size="3" color='#CC403E'><strong>✔️ Classificação de períodos analíticos</strong></font>

<font size="2" color='#66666'></font></div>

```python
def classify_release_period(year: float) -> str:
    """
    Classifica o ano de lançamento em períodos analíticos.
    """
    if pd.isna(year):
        return "unknown"

    year = int(year)

    if year < 2009:
        return "catalogo_historico_pre_2009"
    if 2009 <= year <= 2015:
        return "streaming_early_2009_2015"
    if 2016 <= year <= 2020:
        return "streaming_growth_2016_2020"
    if 2021 <= year <= 2023:
        return "streaming_mature_2021_2023"
    if 2024 <= year <= 2025:
        return "recent_releases_2024_2025"

    return "out_of_scope"


df_silver["periodo_analitico"] = df_silver["release_year"].apply(classify_release_period)

period_summary = (
    df_silver
    .groupby("periodo_analitico", dropna=False)
    .agg(
        track_count=("track_id", "count"),
        min_year=("release_year", "min"),
        max_year=("release_year", "max"),
        avg_track_popularity=("track_popularity", "mean"),
        median_track_popularity=("track_popularity", "median"),
    )
    .reset_index()
    .sort_values("min_year")
)

period_summary
```

<font size="3" color='#CC403E'><strong>✔️ Tratamento controlado de valores ausentes</strong></font>

<font size="2" color='#66666'></font></div>

```python
# Flags de ausência
df_silver["artist_genres_missing_flag"] = df_silver["artist_genres"].isna()
df_silver["artist_name_missing_flag"] = df_silver["artist_name"].isna()

# Tratamento explícito de gênero ausente
df_silver["artist_genres_clean"] = (
    df_silver["artist_genres"]
    .fillna("sem_genero_informado")
    .str.strip()
)

# Tratamento explícito de artista ausente
df_silver["artist_name_clean"] = (
    df_silver["artist_name"]
    .fillna("artista_nao_informado")
    .str.strip()
)

missing_treatment_report = pd.DataFrame(
    {
        "metric": [
            "total_rows",
            "artist_genres_missing_count",
            "artist_genres_missing_pct",
            "artist_name_missing_count",
            "artist_name_missing_pct",
            "artist_genres_clean_missing_count",
            "artist_name_clean_missing_count",
        ],
        "value": [
            len(df_silver),
            df_silver["artist_genres_missing_flag"].sum(),
            round(df_silver["artist_genres_missing_flag"].mean() * 100, 2),
            df_silver["artist_name_missing_flag"].sum(),
            round(df_silver["artist_name_missing_flag"].mean() * 100, 2),
            df_silver["artist_genres_clean"].isna().sum(),
            df_silver["artist_name_clean"].isna().sum(),
        ],
    }
)

missing_treatment_report
```

<div style="background-color:#f3f2ee">
    
<font size="4" color='#CC403E'><strong>📌 12. Marcação de possíveis duplicados textuais</strong></font>

<font size="2" color='#66666'>

>- Marcar possíveis duplicidades textuais sem remover registros da base.

>- **Ações realizadas**
    - Será criada uma chave textual baseada em:
        - *`track_name + artist_name_clean + album_name`*
        - *Também serão criadas variáveis auxiliares para identificar grupos com mais de um registro.*

>- **Justificativa técnica**
    - No Notebook 01 foram identificados casos em que músicas possuem o mesmo nome, artista e álbum, mas `track_id` diferente.
    - Esses casos podem representar versões, relançamentos ou registros alternativos.

>- **Resultado esperado**
    - Marcar os registros potencialmente duplicados para futura decisão sobre visão canônica.

---
</font></div>

<div style="background-color:#f3f2ee">
    
<font size="3" color='#CC403E'><strong>✔️ Marcação de possíveis duplicados textuais</strong></font>

<font size="2" color='#66666'></font></div>

```python
canonical_key_cols = ["track_name", "artist_name_clean", "album_name"]

df_silver["canonical_track_key"] = (
    df_silver["track_name"].fillna("").str.lower().str.strip()
    + " | "
    + df_silver["artist_name_clean"].fillna("").str.lower().str.strip()
    + " | "
    + df_silver["album_name"].fillna("").str.lower().str.strip()
)

canonical_group_size = (
    df_silver
    .groupby("canonical_track_key")["track_id"]
    .transform("count")
)

df_silver["canonical_group_size"] = canonical_group_size
df_silver["possible_text_duplicate_flag"] = df_silver["canonical_group_size"] > 1

duplicate_marking_report = pd.DataFrame(
    {
        "metric": [
            "total_rows",
            "rows_possible_text_duplicate",
            "pct_rows_possible_text_duplicate",
            "canonical_groups_total",
            "canonical_groups_with_duplicates",
        ],
        "value": [
            len(df_silver),
            df_silver["possible_text_duplicate_flag"].sum(),
            round(df_silver["possible_text_duplicate_flag"].mean() * 100, 2),
            df_silver["canonical_track_key"].nunique(dropna=True),
            df_silver.loc[
                df_silver["possible_text_duplicate_flag"],
                "canonical_track_key",
            ].nunique(dropna=True),
        ],
    }
)

duplicate_marking_report
```
