## 2. Carregamento dos Artefatos Finais

Nesta etapa serão carregados os principais resultados do projeto:

- `final_fraud_risk_score.parquet`;
- `final_fraud_alerts.parquet`;
- `graph_entity_ranking.parquet`;
- `community_risk_summary.parquet`.

Esses artefatos representam a consolidação das camadas transacional, antifraude, grafo e explicabilidade.

```python
required_files = {
    "final_score": GOLD_DIR / "final_fraud_risk_score.parquet",
    "final_alerts": GOLD_DIR / "final_fraud_alerts.parquet",
    "graph_entity_ranking": GOLD_DIR / "graph_entity_ranking.parquet",
    "community_risk_summary": GOLD_DIR / "community_risk_summary.parquet",
}

for name, path in required_files.items():
    if not path.exists():
        raise FileNotFoundError(
            f"Arquivo obrigatório não encontrado: {path}. "
            "Verifique se os notebooks anteriores foram executados."
        )

final_score = pd.read_parquet(required_files["final_score"])
final_alerts = pd.read_parquet(required_files["final_alerts"])
graph_entity_ranking = pd.read_parquet(required_files["graph_entity_ranking"])
community_risk_summary = pd.read_parquet(required_files["community_risk_summary"])

final_score["data_hora"] = pd.to_datetime(final_score["data_hora"])
final_alerts["data_hora"] = pd.to_datetime(final_alerts["data_hora"])

print("Artefatos finais carregados com sucesso.")
print(f"Score final:    {final_score.shape}")
print(f"Alertas finais: {final_alerts.shape}")
```
